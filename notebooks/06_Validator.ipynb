```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

sys.path.append(str(PROJECT_ROOT))

from src.etl.loader import load_all_excels
from src.etl.normaliser import normalize_dataframe
from src.etl.validator import DataValidator
```

```python
RAW = PROJECT_ROOT / "data" / "raw"

datasets = load_all_excels(RAW)
```

```python
validator = DataValidator()

for name, df in datasets.items():

    df = normalize_dataframe(df)

    validator.validate(name, df)
```

```python
validation_report = validator.report()

validation_report.head()
```

```python
OUTPUT = PROJECT_ROOT / "data" / "output"

validation_report.to_csv(
    OUTPUT / "validation_failures.csv",
    index=False
)

print("Validation report saved.")
```
