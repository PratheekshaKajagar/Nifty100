```python
import pandas as pd
from pathlib import Path
from IPython.display import display
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA= PROJECT_ROOT/"data"/"raw"
excel_files = sorted(RAW_DATA.glob("*.xlsx"))

print(f"{len(excel_files)} datasets found.")
```

```python
for file in excel_files:
    print("="*80)
    print(file.name.upper())
    print("="*80)
    xls=pd.ExcelFile(file)
    print("Sheets:")
    for sheet in xls.sheet_names:
        print("•", sheet)
    print()
```

```python
for files in excel_files:
    print("="*80)
    print(file.stem.upper())
    print("="*80)
    df=pd.read_excel(file)
    display(df.head())
    print()
    print(df.info)
    print()
    print(df.describe(include="all"))
    print("\n\n")
```

```python
profile = []

for file in excel_files:

    df = pd.read_excel(file)

    profile.append({

        "Dataset": file.stem,

        "Rows": len(df),

        "Columns": len(df.columns),

        "Missing": df.isna().sum().sum(),

        "Duplicates": df.duplicated().sum(),

        "Memory (KB)": round(df.memory_usage(deep=True).sum()/1024,2)

    })

profile = pd.DataFrame(profile)

display(profile)
```

```python
profile.to_csv(

    PROJECT_ROOT /
    "data" /
    "output" /
    "dataset_profile.csv",

    index=False

)

print("Dataset profile saved.")
```
