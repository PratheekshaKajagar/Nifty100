```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import sqlite3
import pandas as pd

DB_PATH = PROJECT_ROOT / "data" / "nifty100.db"

conn = sqlite3.connect(DB_PATH)
```

```python
query = """
SELECT
    company_id,
    year,
    return_on_equity_pct,
    debt_to_equity
FROM financial_ratios
WHERE
    return_on_equity_pct > 15
AND
    debt_to_equity < 1
ORDER BY
    return_on_equity_pct DESC
"""

screener = pd.read_sql(query, conn)

screener
```

```python
OUTPUT = PROJECT_ROOT / "output"

screener.to_csv(
    OUTPUT / "quality_screener.csv",
    index=False
)

print("quality_screener.csv created successfully!")
```
