```python
RAW_DATA = PROJECT_ROOT / "data" / "raw"

datasets = load_all_excels(RAW_DATA)
```

```python
normalized = {}

for name, df in datasets.items():

    normalized[name] = normalize_dataframe(name,df)
```

```python
for name, df in normalized.items():

    print("="*80)

    print(name.upper())

    print("="*80)

    print(df.head())

    print(df.dtypes)

    print()
```

```python
PROCESSED = PROJECT_ROOT / "data" / "processed"

PROCESSED.mkdir(exist_ok=True)

for name, df in normalized.items():

    df.to_csv(
        PROCESSED / f"{name}.csv",
        index=False
    )

print("All datasets saved successfully.")
```
