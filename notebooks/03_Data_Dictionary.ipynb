# Data Dictionary & Metadata Generation

```python
import pandas as pd
from pathlib import Path
from IPython.display import display
```

```python
PROJECT_ROOT= Path.cwd().parent
RAW_DATA= PROJECT_ROOT/"data"/"raw"
PROCESSED_DATA= PROJECT_ROOT/"data"/"processed"
OUTPUT_DATA= PROJECT_ROOT/"data"/"output"
excel_files= sorted(RAW_DATA.glob("*.xlsx"))
print(f"Datasets found: {len(Excel_files)}")
```

```python
data_dictionary = []

for file in excel_files:

    df = pd.read_excel(file)

    for column in df.columns:

        data_dictionary.append({

            "Dataset": file.stem,

            "Column Name": column,

            "Data Type": str(df[column].dtype),

            "Missing Values": int(df[column].isna().sum()),

            "Unique Values": int(df[column].nunique()),

            "Example Value": (
                str(df[column].dropna().iloc[0])
                if not df[column].dropna().empty
                else "NULL"
            )

        })

data_dictionary = pd.DataFrame(data_dictionary)

display(data_dictionary)
```

```python
data_dictionary.to_csv(
    OUTPUT_DATA/"data_dictionary.csv", index=False
    
)
print("Saved Successfully")
```

```python
primary_keys = []

for file in excel_files:

    df = pd.read_excel(file)

    for column in df.columns:

        if df[column].is_unique:

            primary_keys.append({

                "Dataset": file.stem,

                "Possible Primary Key": column

            })

primary_keys = pd.DataFrame(primary_keys)

display(primary_keys)
```

```python
duplicates = []

for file in excel_files:

    df = pd.read_excel(file)

    duplicates.append({

        "Dataset": file.stem,

        "Duplicate Rows": int(df.duplicated().sum())

    })

duplicates = pd.DataFrame(duplicates)

display(duplicates)
```

```python
primary_keys.to_csv(

    OUTPUT_DATA / "possible_primary_keys.csv",

    index=False

)

duplicates.to_csv(

    OUTPUT_DATA / "duplicate_rows.csv",

    index=False

)

print("✅ Metadata Reports Saved")
```
