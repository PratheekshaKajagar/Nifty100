```python
import os
from pathlib import Path

import pandas as pd
import numpy as np

from IPython.display import display
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA = PROJECT_ROOT /"data"/"raw"
PROCESSED_DATA=PROJECT_ROOT/"data"/"processed"
OUTPUT_DIR =PROJECT_ROOT/"data"/"output"

print("Project Root :", PROJECT_ROOT)
print("Raw Data     :", RAW_DATA)
print("Processed    :", PROCESSED_DATA)
print("Output       :", OUTPUT_DIR)
```

```python
excel_files= sorted(RAW_DATA.glob("*.xlsx"))
print(f"found {len(excel_files)} Excel Files\n")
for file in excel_files:
    print(file.name)
```

```python
summary=[]
for files in excel_files:
    xls= pd.ExcelFile(files)
    summary.append({
        "File Name": files.name,
        "Sheet Names": ", ".join(xls.sheet_names),
        "Number of Sheets": len(xls.sheet_names)
    })
summary=pd.DataFrame(summary)
display(summary)
```

```python
dataset_summary = []

for file in excel_files:

    df = pd.read_excel(file)

    dataset_summary.append({
        "Dataset": file.stem,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    })

dataset_summary = pd.DataFrame(dataset_summary)

display(dataset_summary)
```

```python
for file in excel_files:

    print("="*70)
    print(file.stem.upper())
    print("="*70)

    df = pd.read_excel(file)

    print(df.columns.tolist())

    print()
```

```python
missing_summary = []

for file in excel_files:

    df = pd.read_excel(file)

    missing_summary.append({
        "Dataset": file.stem,
        "Missing Values": int(df.isna().sum().sum())
    })

missing_summary = pd.DataFrame(missing_summary)

display(missing_summary)
```

```python
dataset_summary.to_csv(
    OUTPUT_DIR / "dataset_summary.csv",
    index=False
)

missing_summary.to_csv(
    OUTPUT_DIR / "missing_summary.csv",
    index=False
)

print("Dataset summary saved successfully.")
```
