```python
import sqlite3
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

DB_PATH = PROJECT_ROOT / "data" / "nifty100.db"

conn = sqlite3.connect(DB_PATH)
```

```python
tables = pd.read_sql(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name;
    """,
    conn,
)

tables
```

```python
for table in tables["name"]:

    rows = pd.read_sql(
        f"SELECT COUNT(*) AS rows FROM {table}",
        conn,
    )

    print(f"{table:<20} {rows.iloc[0,0]}")
```

```python
pd.read_sql(
    "PRAGMA foreign_key_check;",
    conn
)
```
