```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import sqlite3
import pandas as pd

DB_PATH = PROJECT_ROOT / "data" / "nifty100.db"
OUTPUT= PROJECT_ROOT / "data" / "output"

conn = sqlite3.connect(DB_PATH)

ratio_df = pd.read_sql(
    "SELECT * FROM financial_ratios",
    conn
)

ratio_df.head()
```

```python
edge_cases = []

for _, row in ratio_df.iterrows():

    if pd.isna(row["interest_coverage"]):
        edge_cases.append(
            f"{row.company_id} {int(row.year)} : Interest Coverage = Debt Free"
        )

    if pd.isna(row["return_on_equity_pct"]):
        edge_cases.append(
            f"{row.company_id} {int(row.year)} : Negative Equity"
        )

    if row["debt_to_equity"] == 0:
        edge_cases.append(
            f"{row.company_id} {int(row.year)} : Debt Free Company"
        )
```

```python
with open(
    OUTPUT / "ratio_edge_cases.log",
    "w"
) as f:

    for line in edge_cases:
        f.write(line + "\n")

print("ratio_edge_cases.log created")
```
